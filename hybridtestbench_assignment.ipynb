# Hybrid-Test-Bench case study

This **Hybrid-Test-Bench consists** of a steel frame equipped with two electro-mechanical actuator + force transducer. The horizontal actuator applies a force $f_h$ [kN] to the beam tip and measures the corresponding displacement $u_h$ [mm]. The vertical actuator applies a displacement $u_v$ [mm] to the beam tip and measures the corresponding reaction force $f_v$ [kN]. The HybridTestBench is utilized to develop **hybrid testing** methods at AU-CAE. If you are interested about this methodology, have a loot at [McCrum & Williams, 2026 "An overview of seismic hybrid testing of engineering structures"](https://www.sciencedirect.com/science/article/pii/S0141029616300748).

Below is a schematic of the setup with main dimensions in [mm] and indication of forces/displacements, which has to be used to produce a finite-element model. For each part of the test bench (frame, supports and specimen) cross-sections are given with main dimensions. All parts are made of steel (elastic modulus $E=210$ kN/mm2), expect for the specimen, which is made in aluminum (elastic modulus $E=70$ kN/mm2)

<img src="assets\test_bench_2dof_CAD.png" alt="Test Bench 1-DoF" height="500">

As you can see, the figure already reports a tentative numbering of nodes (1-12), which you can use to define the finite-element model. The coordinates of each node, shall be expressed w.r.t. coordinate system x-y-z and computed based on the dimensions of the drawing.

The following loadigs are applied to the specimen:

$u_v(t) = 20.0 \sin(4\pi t)$ [mm]

$f_h(t) = 1.0 \sin(2\pi t)$ [kN]

The corresponding $f_v(t)$ and $u_h(t)$ are measured.

The bending stiffness $K_b$ of the specimen defines the proportionality between $u_v$ and $f_v$:

$f_v = K_b u_v$ [kN]

The axial stiffness $K_a$ of the specimen defines the proportionality between $u_h$ and $f_h$:

$f_h = K_a u_h$ [kN]

# Assignment
In this project both DIGITAL and PHYSICAL TWINS are represented by the same finite-element model:
* The PHYSICAL TWIN is simulated considering that actual material for the specimen, which is aluminum (elastic modulus $E=70e3$ kN/mm2).
* The DIGITAL TWIN assumes an initial guess for the material parameters of the specimen, which is close to aluminum (elastic modulus $E=100$ kN/mm2) but not exact.
* A "model calibration" service shall be developed to estimate the correct value for the elastic modulus of the specimen, by comparing the force displacement relationships $f_h = K_a u_h$ and $f_v = K_b u_v$ between DIGITAL TWIN and PHYSICAL TWIN.
* In this project, beam2d elements are sufficient to construct the finite-element model of the selected test bench using YAFEM following the example in the Jupyter notebook file **yafem_example_beam2d.ipynb**.
